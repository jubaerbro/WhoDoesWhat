```python
import matplotlib.pyplot as plt
import numpy as np

categories = ["Spatial", "Role", "Mixed"]
base = [2.014, 1.932, 0.075]
rc_dpo = [16.166, 16.307, 0.192]

x = np.arange(len(categories))
width = 0.35

fig, ax = plt.subplots(figsize=(5.2, 4.2))

bars1 = ax.bar(x - width/2, base, width, label="Base")
bars2 = ax.bar(x + width/2, rc_dpo, width, label="Qwen + RC-DPO")

ax.set_ylabel("Mean margin")
ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.set_title("(a) Mean margin", fontsize=12)
ax.legend(frameon=False)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

for bars in [bars1, bars2]:
    for bar in bars:
        h = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2, h + 0.2, f"{h:.3f}",
                ha="center", va="bottom", fontsize=9)

plt.tight_layout()
plt.savefig("figure4a_margin.pdf", bbox_inches="tight")
plt.savefig("figure4a_margin.png", dpi=300, bbox_inches="tight")
plt.show()
```
